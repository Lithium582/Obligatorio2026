# 01 — Comprensión y calidad de los datos

El objetivo es entender la estructura del dataset antes de tomar decisiones de modelado.

## Definición del problema

- Formular la pregunta predictiva.
- Identificar la variable objetivo.
- Determinar si se trata de clasificación o regresión.
- Elegir y justificar la métrica principal.
- Identificar restricciones prácticas o costos de error.

## Auditoría inicial

- Dimensiones, tipos y significado de las columnas.
- Identificadores y variables no utilizables.
- Duplicados y valores faltantes.
- Valores imposibles o inconsistentes.
- Distribución de la variable objetivo.
- Riesgos iniciales de fuga de información.
